## Engineering Strategy
1. **Market Consensus Probabilities:** Mean & median fair probabilities removing the bookmaker overround.
2. **Market Disagreement (Dispersion):** Standard deviation of odds across bookmakers for Home, Draw, Away.
3. **Odds Spread:** Max odds vs Min odds available in the market.
4. **Odds Drift Dynamics:** Percentage shift from opening quotes to closing prices.
5. **Temporal & Calendar Features:** Month, day of week, kickoff hour, weekend flag.
6. **Strict Leakage Prevention:** Absolutely zero post-kickoff or score features are included in predictor vectors.

## 2. Feature Correlation with Match Outcomes

```python
corr_cols = [
    'mkt_prob_home', 'mkt_prob_draw', 'mkt_prob_away',
    'odds_std_home', 'odds_std_away', 'spread_odds_home',
    'mean_drift_home', 'mean_overround'
]

# Binarize Home Win for correlation test
home_win_binary = (features['target'] == 0).astype(int)
corrs = features[corr_cols].apply(lambda s: s.corr(home_win_binary))
print("Correlation with Home Win (target == 0):")
print(corrs.sort_values(ascending=False).round(4))
```
